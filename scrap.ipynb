```python
# 웹 페이지를 요청하고 응답 데이터를 받을 수 있다.
import requests as req
from bs4 import BeautifulSoup # BeautifulSoup html 해석기

# 요청시 오류 났을 때 아래 문구를 입력해주세요.
request_headers = {'User-Agent': ('Mozilla/5.0 (Windows NT 10.0;Win64; x64)AppleWebKit/537.36 (KHTML, like Gecko) Chrome/71.0.3578.98Safari/537.36'), }

# 특정 사이트에 페이지를 요청. 응답 데이터 반환
res = req.get('https://www.naver.com/', request_headers)

# print(res)


html_doc = """<html><head><title>The Dormouse's story</title></head>
<body>
<p class="title"><b>The Dormouse's story</b></p>

<p class="story">Once upon a time there were three little sisters; and their names were
<a href="http://example.com/elsie" class="sister" id="link1">Elsie</a>,
<a href="http://example.com/lacie" class="sister" id="link2">Lacie</a> and
<a href="http://example.com/tillie" class="sister" id="link3">Tillie</a>;
and they lived at the bottom of a well.</p>

<p class="story">...</p>
"""

# print(html_doc)
# soup을 이용해서 html_doc 문자열 파싱. 그 내용을 soup 변수에 대입
soup = BeautifulSoup(html_doc, 'html.parser')

alist = soup.select('a')

print('== a태그들 리스트 ==')
print(alist)

print('== a태그들 리스트 중 첫 번째 ==')
print(alist[0])

print('== a태그들 리스트 중 첫 번째 텍스트 ==')
print(alist[0].text)

print('== a태그들 리스트 중 첫 번째 ==')
print(alist[0]['href']) # 속성 선택할 땐 대괄호

print('== a태그들 리스트에 있는 텍스트 모두 출력 ==')
for a in alist:
    print(a.text)

print('== a태그들 리스트에 있는 링크 모두 출력 ==')
for link in alist:
    print(link['href'])
```

```python
import requests as req
from bs4 import BeautifulSoup

res = req.get('https://news.naver.com/')
# print(res.text)

soup = BeautifulSoup(res.text,'html.parser')
content = soup.select('.cnf_news_title')

print('== class로 선택, 하나만 텍스트 출력 ==')
print(content[0].text)

print('== class로 선택, for문으로 텍스트 출력 ==')
for c in content:
    print(c.text)

print('== 자식 선택자로 선택, for문으로 텍스트 출력 ==')
content2 = soup.select('.cnf_news_area > strong')
for c in content2:
    print(c.text)
```
